```sql
-- CASCADE drops the object even if it contains other sub objects otherwise it will throw an error
-- CATALOG is the top level object
DROP CATALOG IF EXISTS multi_flow_SDP CASCADE;  
CREATE CATALOG IF NOT EXISTS multi_flow_SDP;
```

```sql
-- Create three catalog schemas for multi_flow_sdp Section

CREATE SCHEMA IF NOT EXISTS multi_flow_sdp.01_bronze_orders;
CREATE SCHEMA IF NOT EXISTS multi_flow_sdp.02_silver_orders;
CREATE SCHEMA IF NOT EXISTS multi_flow_sdp.03_gold_orders;
```

```sql
-- Create three volumes for three different companies

CREATE VOLUME IF NOT EXISTS multi_flow_sdp.01_bronze_orders.bright_home_orders;
CREATE VOLUME IF NOT EXISTS multi_flow_sdp.01_bronze_orders.lumina_sports_orders;
CREATE VOLUME IF NOT EXISTS multi_flow_sdp.01_bronze_orders.northstar_outfitters_orders;
```
